## Step 1: Gold Standard Annotation

I chose 20 news articles from my Meta dataset in Assignment 1. I'll read through each article and label three types of entities:

- **PERSON:** People's names, including full names and last names.
- **ORG:** Companies, organizations, and government agencies.
- **GPE:** Countries, cities, and states.

### Annotation Rules

1. I'll only label names that actually appear in the article. Words like "the company" or "the CEO" won't count.
2. For names like "Meta's", I'll only mark "Meta" and leave out the 's.
3. Stock symbols like "NASDAQ:META" won't count as company names.
4. If the same name appears more than once, I'll mark it every time.
5. I'll keep names the way they're written in the article, even if they're shortened or abbreviated.

### Ambiguous Cases

1. **Meta vs. Meta's:** When an organization name has 's at the end, I only label the organization name itself. For example, I label "Meta" instead of "Meta's" because the 's just shows possession.

2. **Meta Verified vs. Meta:** "Meta Verified" is a product name, so I don't label the whole phrase as ORG. I only label "Meta" when it refers to the company.

```python
import re
import pandas as pd

text = articles.loc[0, "text"]

entity_labels = {
    "Megha Reddy": "PERSON",
    "Meta": "ORG",
    "Instagram": "ORG",
    "Facebook": "ORG",
    "Twitter": "ORG",
    "India": "GPE",
    "Brazil": "GPE",
    "Canada": "GPE",
    "UK": "GPE",
    "Australia": "GPE",
    "New Zealand": "GPE",
    "US": "GPE",
    "California": "GPE"
}

annotations = []

for entity, label in entity_labels.items():
    pattern = r"(?<!\w)" + re.escape(entity) + r"(?!\w)"
    for match in re.finditer(pattern, text):
        annotations.append({
            "article_id": 1,
            "entity": match.group(),
            "label": label,
            "start": match.start(),
            "end": match.end()
        })

annotations = sorted(annotations, key=lambda x: x["start"])

annotation_df = pd.DataFrame(annotations)

display(annotation_df)
print("Total candidate annotations:", len(annotation_df))
```

```python
gold[1] = [
    (row["entity"], row["label"], row["start"], row["end"])
    for _, row in annotation_df.iterrows()
]

print("Article 1 saved:", len(gold[1]))
print(gold[1])
```

## Step 4: Evaluation

I compared spaCy and Llama 3.2 with my gold standard using precision, recall, and F1.

I used exact matching for both entity names and labels. I ignored differences in capitalization and extra spaces, but shortened names like "Zuckerberg" and "Mark Zuckerberg" were treated as different entities. I also counted repeated mentions, since the same name can appear more than once in an article.


```python
from collections import Counter

def get_counts(predicted, actual, label):
    p = Counter()
    g = Counter()

    for name, tag in predicted:
        if tag == label:
            p[name.lower().strip()] += 1

    for name, tag in actual:
        if tag == label:
            g[name.lower().strip()] += 1

    correct = sum((p & g).values())

    return correct, sum(p.values()), sum(g.values())


def evaluate(results):
    rows = []
    total_correct = 0
    total_predicted = 0
    total_actual = 0

    for label in ["PERSON", "ORG", "GPE"]:
        correct = 0
        predicted = 0
        actual = 0

        for article_id in range(1, 21):
            gold_entities = [(x[0], x[1]) for x in gold[article_id]]
            model_entities = [(x[0], x[1]) for x in results[article_id]]

            c, p, g = get_counts(model_entities, gold_entities, label)
            correct += c
            predicted += p
            actual += g

        precision = correct / predicted if predicted else 0
        recall = correct / actual if actual else 0
        f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0

        rows.append([label, precision, recall, f1])

        total_correct += correct
        total_predicted += predicted
        total_actual += actual

    precision = total_correct / total_predicted if total_predicted else 0
    recall = total_correct / total_actual if total_actual else 0
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0

    rows.append(["Overall", precision, recall, f1])

    return pd.DataFrame(rows, columns=["Label", "Precision", "Recall", "F1"])
```

```python
spacy_scores = evaluate(spacy_results)
llm_scores = evaluate(llm_results)

print("spaCy")
display(spacy_scores.round(3))

print("Llama 3.2")
display(llm_scores.round(3))
```

```python
from collections import Counter

def error_types(results, name):
    rows = []
    for aid in range(1, 21):
        g = Counter((x[0].lower().strip(), x[1]) for x in gold[aid])
        p = Counter((x[0].lower().strip(), x[1]) for x in results[aid])
        g_lab = {t: l for t, l in g}
        p_txt = {t for t, l in p}
        tok = lambda s: set(s.split())
        for (t, l), n in (p - g).items():
            if t in g_lab and g_lab[t] != l: kind = "wrong label"
            elif (t, l) in g: kind = "extra mentions"
            elif any(tok(t) & tok(gt) for gt in g_lab): kind = "boundary"
            else: kind = "not in gold"
            rows.append((name, aid, "FP", kind, t, l, n))
        for (t, l), n in (g - p).items():
            if (t, l) in p: kind = "fewer mentions"
            elif t in p_txt: kind = "wrong label"
            elif any(tok(t) & tok(pt) for pt in p_txt): kind = "boundary"
            else: kind = "missed"
            rows.append((name, aid, "FN", kind, t, l, n))
    return rows

err = pd.DataFrame(error_types(spacy_results, "spaCy") + error_types(llm_results, "Llama"),
                   columns=["model", "article", "error", "kind", "entity", "label", "n"])
display(err.groupby(["model", "error", "kind"])["n"].sum().unstack(0).fillna(0).astype(int))
for (m, e, k), grp in err.groupby(["model", "error", "kind"]):
    print(m, e, k, "->", list(zip(grp["entity"], grp["label"], grp["article"]))[:5])
```

### Results and Discussion

Llama 3.2 had higher precision (0.680), but spaCy had higher recall (0.611). Their F1 scores were almost the same, around 0.61.

Llama 3.2 was better at finding people's names, but it missed quite a few organizations. spaCy found more names overall, but some were labeled incorrectly. For example, it labeled "Twitter" as PERSON in Article 1.

Some names also didn't match because they were written differently. For example, spaCy identified "Meta Verified," while I only labeled "Meta" in my gold standard. Since I used exact matching, this counted as an error.

### Model Differences

The two models made different kinds of mistakes.

spaCy usually found the right words but gave them the wrong label or the wrong boundaries. It labeled "Twitter" and "Meta" as PERSON in Article 1, and "Messenger" as PERSON in Article 3. It also included extra words, like "Meta Verified" and "WhatsApp Channels", when I only labeled "Meta" and "WhatsApp". I think this happens because the small model mostly looks at capital letters and nearby words.

Llama 3.2 had fewer label mistakes, but it missed a lot of entities. In Article 1 it skipped Instagram, Facebook and Twitter, and it often listed a name only once even when it appeared several times. This explains its low ORG recall (0.456). Splitting the articles into small chunks probably made this worse. It also made some strange mistakes, like labeling "November" and "September" as GPE in Article 7.

There was also one case where the LLM and I just disagreed. It labeled "EU" as GPE, but I labeled it as ORG. Both could be argued.

### Which one would I use?

For 20 articles, both models had almost the same F1 (0.607 vs 0.608), so either one works.

For 10 million articles I would use spaCy. Llama was slow on my laptop. I had to stop some runs and split the articles to make them finish. Even at 10 seconds per article, 10 million articles would take about 3 years on one computer. spaCy can process thousands of articles per minute. An LLM API would be faster, but it charges per token, so the cost would be very high. Llama also returned broken JSON once (Article 18), which would be a bigger problem at that scale.

## Step 5: Wikidata Entity Linking

```python
from collections import Counter

org_counts = Counter()

for entities in spacy_results.values():
    for entity, label, start, end in entities:
        if label == "ORG":
            org_counts[entity] += 1

top_orgs = org_counts.most_common(20)

display(pd.DataFrame(top_orgs, columns=["Organization", "Count"]))
```

```python
import requests

def search_wikidata(name):
    url = "https://www.wikidata.org/w/api.php"

    params = {
        "action": "wbsearchentities",
        "search": name,
        "language": "en",
        "format": "json",
        "limit": 1
    }

    response = requests.get(
        url,
        params=params,
        headers={"User-Agent": "APAN5400-StudentProject/1.0"},
        timeout=15
    )

    response.raise_for_status()
    results = response.json()["search"]

    if results:
        return results[0]["id"], results[0]["label"]

    return None, None
```

```python
links = []

for name, count in top_orgs:
    try:
        wikidata_id, wikidata_name = search_wikidata(name)
        links.append([name, count, wikidata_id, wikidata_name])
        print(name, "done")
    except Exception as e:
        links.append([name, count, None, None])
        print(name, "failed:", str(e))

link_df = pd.DataFrame(
    links,
    columns=["Organization", "Count", "Wikidata ID", "Wikidata Name"]
)

display(link_df)
```

### Wikidata Results

Out of the 20 organization names, **9 were linked correctly, 11 were linked to the wrong entity, and 0 could not be linked.**

- **Correct example:** "Google" → Q95 (Google), the company mentioned in the articles.
- **Wrong example:** "Meta" → Q238629, which is the Meta Department in Colombia, not the company Meta Platforms. This is the most frequent ORG in my articles (61 mentions), and it still linked to the wrong entity because "Meta" is also a place name and I only took the first search result.
- **Could not be linked:** none. Every name returned at least one result, because the Wikidata search does fuzzy matching. The "429 Too Many Requests" errors were only rate limits. They worked when I retried with a pause, so they don't count as unlinkable.

Some mistakes happened because the same name can refer to different things. Wikidata linked "Sama" to a place in Spain instead of the company in the Kenya article. It also linked "Channels" to a family name and "Facebook and Instagram" to a news article.
