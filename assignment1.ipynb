The free account only has 100 calls, so it will be saved as sec_meta.json after the first pull. The 50 documents retrieved mainly consist of Form 4 and Form 144, which mainly contain declarations of stock buying and selling by company insiders.

```python
from sec_api import QueryApi

query_api = QueryApi(api_key=api_key)

query = {
    "query": "ticker:META",
    "from": "0",
    "size": "50",
    "sort": [{"filedAt": {"order": "desc"}}]
}

response = query_api.get_filings(query)

with open("data/sec_meta.json", "w") as f:
    json.dump(response["filings"], f)

print(len(response["filings"]))
```

```python
with open("data/sec_meta.json", "r") as f:
    filings = json.load(f)

sec = pd.DataFrame(filings)
print(sec.shape)
print(sec.columns.tolist())
sec[["formType", "description", "filedAt", "linkToFilingDetails"]]
```

The data fields on both sides are not exactly the same, so I made some simple adjustments:
title: There is no direct title in the SEC data, so I put the company name and form type together, such as "Meta Platforms, Inc. Form 4".
text: The Query API mainly returns the basic information of the filing without the complete text, so description is used here first. This is relatively simple, but the SEC text will be shorter, and the information may not be as rich as news data when doing text analysis later.

```python
news_df = pd.DataFrame({
    "source": "news",
    "title": df_unique["title"],
    "text": df_unique["text"],
    "url": df_unique["url"],
    "date": df_unique["published"]
})

sec_df = pd.DataFrame({
    "source": "sec",
    "title": sec["companyName"] + " Form " + sec["formType"],
    "text": sec["description"],
    "url": sec["linkToFilingDetails"],
    "date": pd.to_datetime(sec["filedAt"], utc=True)
})

combined = pd.concat([news_df, sec_df], ignore_index=True)
print(combined.shape)
print(combined["source"].value_counts())
```

### Initial observations

There is a pretty big difference between the news and SEC data. The news articles contain over 2.5 million words, while the SEC data only has 515. Most of the common words in the SEC data come from repeated filing descriptions, such as "Statement of changes in beneficial ownership."

There are also a few problems with simply splitting the text by spaces:

- Uppercase and lowercase words are counted separately, such as `the` and `The`.
- Punctuation stays attached to words, so `Meta` and `Meta,` are treated as different words.
- Symbols such as `-` are also counted as words.
- Most of the top words are common words like `the`, `and`, and `of`, which do not tell us much about what the articles are actually about.
- Because of this, the 214,449 unique word count is probably higher than the actual number of unique words.
- I also noticed that the dataset includes some articles in other languages, such as Portuguese

```python
from collections import Counter

for src in ["news", "sec"]:
    words = []
    for t in combined[combined["source"] == src]["text"]:
        words.extend(t.split())
    print(src)
    print("total words:", len(words))
    print("unique words:", len(set(words)))
    print(Counter(words).most_common(25))
    print()
```

I set 200 characters as "too short". 200 characters only contains about 30 to 40 English words, which is less than a normal paragraph. Content below this length is basically impossible to analyze. As a result, 439 News articles were below 200, and all 50 SEC articles were below 200, with no completely empty documents on either side.

```python
cutoff = 200

for src in ["news", "sec"]:
    part = combined[combined["source"] == src]
    empty = part[part["text"].str.strip() == ""]
    short = part[part["text"].str.len() < cutoff]
    print(src, "empty:", len(empty), "shorter than", cutoff, "chars:", len(short))
```

```python
news_short = combined[(combined["source"] == "news") & (combined["text"].str.len() < cutoff)]
news_short = news_short.assign(length=news_short["text"].str.len()).sort_values("length")

for i, row in news_short.head(10).iterrows():
    print(row["length"], row["title"])
    print(repr(row["text"]))
    print()
```
